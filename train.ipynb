```python
import pandas as pd
import json

# Load your CSV file  # Provide your CSV file path here
df = pd.read_csv(csv_file)

# Base video path
base_video_path = '/content/drive/MyDrive/Dataset/videos/'

# Initialize the structure for the JSON output
output_data = {
    "name": "My Config",
    "data": []
}

# Iterate over unique video names in the CSV
for video_name in df['PATH_TO_VIDEO'].unique():
    # Filter rows related to the current video name
    video_data = df[df['PATH_TO_VIDEO'] == video_name]

    # Create the video entry
    video_entry = {
        "video_path": base_video_path + video_name,
        "num_frames": len(video_data),
        "data": []
    }

    # Iterate over the rows for this video and create frame data
    for index, row in video_data.iterrows():
        frame_entry = {
            "frame_index": (index % 10) + 1,  # Assuming video is divided into 10-frame sets, adjust frame index accordingly
            "prompt": row['CAPTION']
        }
        video_entry['data'].append(frame_entry)

    # Add the video entry to the main data
    output_data["data"].append(video_entry)

# Save the output data as a JSON file
output_json_file = '/content/my_config.json'  # Specify the output JSON file name
with open(output_json_file, 'w') as json_file:
    json.dump(output_data, json_file, indent=4)

print(f"JSON file saved as {output_json_file}")
```
